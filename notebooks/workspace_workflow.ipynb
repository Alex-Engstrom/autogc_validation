# Workspace Workflow

Two-phase pipeline for monthly AutoGC validation:

1. **`create_workspace`** — creates the folder structure
2. Copy zipped files into `temp/`
3. **`process_workspace`** — unzips, moves, and sorts files

```python
import logging
from pathlib import Path
from autogc_validation.workspace import create_workspace, process_workspace

logging.basicConfig(level=logging.INFO)
logger = logging.getLogger(__name__)
```

## Configuration

```python
root_dir = Path(r"C:\path\to\validation")
sitename = "RB"
year = 2026
month = 1
```

## Phase 1: Create folder structure

```python
result = create_workspace(root_dir, sitename, year, month)
print(f"Workspace created: {result.base_dir}")
print(f"Copy zipped files into: {result.base_dir / 'temp'}")
```

## Copy files

**Stop here.** Copy zipped `.zip` files from the network location into the `temp/` folder printed above, then continue.

## Phase 2: Unzip, move, and sort files

```python
result = process_workspace(result.base_dir)

print(f"Steps completed: {result.steps_completed}")
print(f"Errors: {result.errors}")
if result.dat_summary:
    print(f"DAT files: {result.dat_summary['found'][0]} found, {result.dat_summary['copied'][0]} copied")
if result.tx1_summary:
    print(f"TX1 files: {result.tx1_summary['found'][0]} found, {result.tx1_summary['copied'][0]} copied")
if result.week_counts:
    print(f"Week counts: {result.week_counts}")
```
