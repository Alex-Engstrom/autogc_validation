```python
import re
from typing import Union
import shutil
from pathlib import Path

def move_MAX_txt(input_dir: Union[Path, str], output_dir: Union[Path, str] = None)-> tuple:
    """ Use to manually move txt files from one folder to another."""
    input_dir = Path(input_dir)
    counter = 0
    if not output_dir:
        output_dir = input_dir / "MAX upload"
        output_dir.mkdir(exist_ok=True)
    pattern = re.compile(r"[A-Za-z]{4}\d{2}[A-Za-z]\.tx1")
    matches = [file for file in input_dir.iterdir() if pattern.match(file.name)]
    for file in matches:
        shutil.copy2(file, output_dir)
        counter += 1
    print("files moved", counter)
    return (counter, output_dir)
input_dir = Path(r"D:\autogc_validation\validation\EQ\EQ202504v1\FINAL\week 4")
counts, path = move_MAX_txt(input_dir)
path
```
